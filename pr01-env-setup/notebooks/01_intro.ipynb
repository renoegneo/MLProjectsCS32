## Визуализация

```python
sns.pairplot(df, hue='species', height=1.7)
plt.show()
```

```python
sns.countplot(data=df, x='species')
plt.title('Количество пингвинов по видам')
plt.show()
```

## Data dictionary

| Колонка           | Значение                                   | Тип   | Единицы |
|-------------------|--------------------------------------------|-------|---------|
| species           | Вид пингвина (Adelie / Chinstrap / Gentoo) | str   | —       |
| island            | Остров, где сделан промер                  | str   | —       |
| culmen_length_mm  | Длина клюва (culmen)                       | float | мм      |
| culmen_depth_mm   | Высота клюва                               | float | мм      |
| flipper_length_mm | Длина ласта                                | float | мм      |
| body_mass_g       | Масса тела                                 | float | граммы  |
| sex               | Пол особи                                  | str   | —       |

## Выводы

Датасет содержит 344 строки и 7 колонок (4 числовых, 3 категориальных).
Пропуски есть в культен-промерах и весе,
а также в поле sex (10 пропусков). Три вида пингвинов представлены неравномерно:
Adelie (~150) заметно больше, чем Chinstrap (~68) и Gentoo (~124).
На pairplot виден хороший визуальный разброс: Gentoo чётко отделяется по flipper_length_mm
и body_mass_g от двух других видов.
